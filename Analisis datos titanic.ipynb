```python
df = df.drop('Cabin', axis=1)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())  
```

```python
df = df.dropna(subset=['Embarked'])
```

```python
df.isnull().sum()
```

```python
import seaborn as sns  

sns.barplot(x='Sex', y='Survived', data=df)
```

```python
import plotly.express as px 

fig = px.bar(
    df.groupby(['Pclass','Sex'], as_index=False)['Survived'].mean(),
    x='Pclass', 
    y='Survived', 
    color='Sex',
    barmode='group',
    text_auto='.1%',
    title='Survival Rate by Sex and Pclass',
    template='plotly_white',
    color_discrete_sequence=['#2E86AB','#E84855']
)
fig.show()  
```

```python
import numpy as np
import plotly.express as px

df['AgeGroup'] = np.where(df['Age'] < 18, 'Child', 'Adult (18+)')
resume_child = df.groupby(['Pclass', 'AgeGroup'] , as_index=False)['Survived'].mean()  
print(resume_child)

fig = px.bar(
    resume_child,
    x='Pclass',
    y='Survived',
    color='AgeGroup',
    barmode='group',
    text_auto='.1%',
    title='Survival Rate by Age Group and Pclass',
    template='plotly_white',
    color_discrete_sequence=["#32AB70","#B6444D"]
)   
fig.show()  
```
